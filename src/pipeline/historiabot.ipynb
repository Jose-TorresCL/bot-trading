```python
import pandas as pd

# Cargar resultados
df = pd.read_csv("results/backtest_equity.csv", parse_dates=["timestamp"])
trades = pd.read_csv("results/backtest_trades.csv")

# Métricas clave
total_trades = len(trades)
winrate = (trades.pnl > 0).mean()
profit_factor = trades.pnl[trades.pnl > 0].sum() / abs(trades.pnl[trades.pnl < 0].sum())
expectancy = trades.pnl.mean()
drawdown_max = (df.equity / df.equity.cummax() - 1).min()

print(f"Total trades: {total_trades}")
print(f"Winrate: {winrate:.2%}")
print(f"Profit Factor: {profit_factor:.2f}")
print(f"Expectancy: {expectancy:.2f}")
print(f"Drawdown máximo: {drawdown_max:.2%}")
```

```python
## 🧠 Aprendizajes clave
- El orden y la trazabilidad son tan importantes como la lógica de trading.
- Backtesting ≠ realidad → validar en entornos operativos.
- Automatizar análisis y documentar decisiones mejora eficiencia.
- Git y notebooks son herramientas clave para escalar con control.
```

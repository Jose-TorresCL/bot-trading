```python
# Este notebook está especializado en el análisis avanzado de resultados de Paper Trading.
# Está preparado para detectar y analizar automáticamente las columnas disponibles.
# Si en el futuro agregas nuevas métricas o columnas, solo actualiza los resúmenes y visualizaciones.
# Si algún archivo falta, el notebook lo notificará y continuará con los análisis posibles.
## Análisis avanzado de operaciones individuales en modo Paper Trading

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import json
import os

# Rutas relativas desde notebooks/
archivos = {
    "papertrading_BTCUSDT.csv": "../data/papertrading/papertrading_BTCUSDT.csv",
    "estado_papertrading.json": "../data/papertrading/estado_papertrading.json",
    "registro_decisiones.json": "../data/papertrading/registro_decisiones.json",
    "registro_operaciones.json": "../data/papertrading/registro_operaciones.json"
}

print("Verificando archivos y rutas desde la carpeta 'notebooks':\n")
for nombre, ruta in archivos.items():
    if os.path.exists(ruta):
        print(f"✅ {nombre}: encontrado en {ruta}")
    else:
        print(f"❌ {nombre}: NO encontrado en {ruta}")

def cargar_csv(ruta):
    if os.path.exists(ruta):
        return pd.read_csv(ruta)
    else:
        print(f"⚠️ El archivo {ruta} no existe.")
        return pd.DataFrame()

def cargar_json(ruta):
    if os.path.exists(ruta):
        with open(ruta, encoding="utf-8") as f:
            return json.load(f)
    else:
        print(f"⚠️ El archivo {ruta} no existe.")
        return {}

def cargar_json_lines(ruta):
    datos = []
    errores = 0
    if os.path.exists(ruta):
        with open(ruta, encoding="utf-8") as f:
            for line in f:
                line = line.strip()
                if line:
                    try:
                        datos.append(json.loads(line))
                    except Exception as e:
                        errores += 1
                        if errores <= 5:
                            print(f"⚠️ Error en línea JSON: {e} | Línea: {line}")
        if errores > 0:
            print(f"⚠️ Total de líneas con error en {ruta}: {errores}")
    else:
        print(f"⚠️ El archivo {ruta} no existe.")
    return datos

# Cargar datos principales
df_paper = cargar_csv(archivos["papertrading_BTCUSDT.csv"])
estado_paper = cargar_json(archivos["estado_papertrading.json"])
decisiones = cargar_json_lines(archivos["registro_decisiones.json"])
df_decisiones = pd.DataFrame(decisiones)
operaciones = cargar_json_lines(archivos["registro_operaciones.json"])
df_operaciones = pd.DataFrame(operaciones)

# Visualización básica
plt.figure(figsize=(10, 5))
plt.plot(df_paper["timestamp"], df_paper["saldo_virtual"])
plt.xticks(rotation=45)
plt.title("Evolución del saldo virtual (Paper Trading)")
plt.xlabel("Fecha")
plt.ylabel("Saldo USDT")
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(df_paper["timestamp"], df_paper["cantidad_btc"], color="orange")
plt.xticks(rotation=45)
plt.title("Cantidad de BTC en Paper Trading")
plt.xlabel("Fecha")
plt.ylabel("Cantidad BTC")
plt.show()

print("Últimas operaciones simuladas:")
display(df_paper.tail(10))

print("Estado actual del paper trading:")
print(estado_paper)

if not df_operaciones.empty:
    print("Resumen de operaciones registradas:")
    display(df_operaciones.tail(10))

if not df_decisiones.empty:
    print("Últimas decisiones tomadas por el bot:")
    display(df_decisiones.tail(10))

# Resumen estadístico de operaciones
if not df_operaciones.empty:
    resumen = df_operaciones.agg({
        "ganancia": ["mean", "max", "std"],
        "precio": ["mean", "max", "std"],
        "tipo": "count"
    })
    print("Resumen estadístico de operaciones:")
    display(resumen)

    # Exportar operaciones robustas (ganancia > mediana, duración < mediana si existe)
    if "ganancia" in df_operaciones and "duracion" in df_operaciones:
        robustas = df_operaciones[
            (df_operaciones["ganancia"] > df_operaciones["ganancia"].median()) &
            (df_operaciones["duracion"] < df_operaciones["duracion"].median())
        ]
        robustas.to_csv("../data/papertrading/operaciones_robustas_papertrading.csv", index=False)
        print("Operaciones robustas exportadas a ../data/papertrading/operaciones_robustas_papertrading.csv")
        display(robustas)

    # Exportar mejor operación
    mejor = df_operaciones.loc[df_operaciones["ganancia"].idxmax()]
    mejor_dict = mejor.to_dict()
    with open("../data/papertrading/mejor_operacion_papertrading.json", "w") as f:
        json.dump(mejor_dict, f, indent=4)
    print("Mejor operación exportada a ../data/papertrading/mejor_operacion_papertrading.json")
    display(mejor)

# Gráficos avanzados si hay suficientes datos
if not df_operaciones.empty and "ganancia" in df_operaciones and "duracion" in df_operaciones:
    plt.figure(figsize=(10, 5))
    sns.boxplot(y="ganancia", data=df_operaciones)
    plt.title("Distribución de Ganancia en Paper Trading")
    plt.show()

    plt.figure(figsize=(10, 5))
    sns.scatterplot(x="duracion", y="ganancia", data=df_operaciones)
    plt.title("Ganancia vs Duración en Paper Trading")
    plt.show()
```
